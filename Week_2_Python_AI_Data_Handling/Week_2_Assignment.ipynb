## Task 1 — NumPy Fundamentals

NumPy stands for Numerical Python. It is one of the most important libraries used in AI and Data Science because it makes it easy to work with large amounts of numerical data very fast. It provides arrays which are more efficient than normal Python lists, and it has a lot of built-in functions for doing math operations.

In this task I will create 1D and 2D arrays, explore indexing and slicing, check shape and dimensions, do basic math operations, and use statistical functions like mean, max, min and sum.

```python
import numpy as np

# creating a 1D array with student Math marks
marks = np.array([85, 72, 90, 55, 40, 78, 95, 35, 62, 88])
print("1D Array (Math Marks):")
print(marks)
```

```python
# creating a 2D array - rows are students, columns are Math, Science, English
marks_2d = np.array([
    [85, 78, 82],
    [72, 68, 75],
    [90, 95, 88],
    [55, 60, 65],
    [40, 45, 50],
    [78, 80, 72],
    [95, 92, 90],
    [35, 40, 38],
    [62, 58, 70],
    [88, 85, 80]
])
print("2D Array (Math, Science, English for 10 students):")
print(marks_2d)
```

```python
# array indexing
print("Array Indexing:")
print("First student mark (index 0):", marks[0])
print("Last student mark (index -1):", marks[-1])
print("Third student Science marks (row 2, col 1):", marks_2d[2][1])
```

```python
# array slicing
print("Array Slicing:")
print("Students 2 to 5 marks:", marks[2:6])
print("First 3 students all subjects:")
print(marks_2d[0:3])
print("Math column for all students:", marks_2d[:, 0])
```

```python
# shape and dimensions
print("Shape of 1D array:", marks.shape)
print("Shape of 2D array:", marks_2d.shape)
print("Dimensions in 1D array:", marks.ndim)
print("Dimensions in 2D array:", marks_2d.ndim)
```

```python
# basic math operations
print("Math Operations on marks array:")
print("Original:", marks)
print("Add 5 to each:", marks + 5)
print("Subtract 5:", marks - 5)
print("Multiply by 2:", marks * 2)
print("Divide by 2:", marks / 2)
```

```python
# statistical functions
print("Statistical Functions:")
print("Mean (average):", marks.mean())
print("Max (highest):", marks.max())
print("Min (lowest):", marks.min())
print("Sum (total):", marks.sum())
```

## Task 3 — Data Cleaning

In real life, data is never perfect. It usually has missing values, duplicate records, and sometimes wrong data types. This is a very important part of working with data because if we don't clean it first, our analysis will give wrong results.

In this task I will load a messy dataset and clean it step by step — removing duplicates, filling missing values, and fixing wrong data types.

```python
import pandas as pd

# loading the messy dataset
df_messy = pd.read_csv("messy_student_data.csv")

print("Dataset BEFORE Cleaning:")
df_messy
```

```python
# checking missing values
print("Missing values in each column:")
print(df_messy.isnull().sum())
print("Total missing values:", df_messy.isnull().sum().sum())
```

```python
# checking duplicate rows
print("Number of duplicate rows:", df_messy.duplicated().sum())
print()
print("Duplicate rows:")
df_messy[df_messy.duplicated(keep=False)]
```

```python
# data types before cleaning
print("Data types before cleaning:")
print(df_messy.dtypes)
```

```python
# step 1 - remove duplicate rows (keep first occurrence)
df_clean = df_messy.drop_duplicates()
print("After removing duplicates, rows remaining:", len(df_clean))
```

```python
# step 2 - fix Age column (some values like 'twenty' are text, not numbers)
# pd.to_numeric with errors='coerce' converts text to NaN so we can handle it
df_clean["Age"] = pd.to_numeric(df_clean["Age"], errors="coerce")
print("Age column after fixing wrong data type:")
print(df_clean["Age"])
```

```python
# step 3 - fill missing values
# filling marks with column average because average is a reasonable estimate
# filling age with median because median handles outliers better than mean
# filling Name with 'Unknown' since we can't guess a name
df_clean["Math"] = df_clean["Math"].fillna(df_clean["Math"].mean())
df_clean["Science"] = df_clean["Science"].fillna(df_clean["Science"].mean())
df_clean["English"] = df_clean["English"].fillna(df_clean["English"].mean())
df_clean["Attendance"] = df_clean["Attendance"].fillna(df_clean["Attendance"].mean())
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())
df_clean["Name"] = df_clean["Name"].fillna("Unknown")

# rounding float values and converting age to integer
df_clean["Math"] = df_clean["Math"].round(1)
df_clean["Science"] = df_clean["Science"].round(1)
df_clean["English"] = df_clean["English"].round(1)
df_clean["Attendance"] = df_clean["Attendance"].round(1)
df_clean["Age"] = df_clean["Age"].astype(int)

print("Missing values filled successfully.")
```

```python
# dataset AFTER cleaning
print("Dataset AFTER Cleaning:")
df_clean
```

```python
# verification
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
print()
print("Duplicate rows after cleaning:", df_clean.duplicated().sum())
print()
print("Data types after cleaning:")
print(df_clean.dtypes)
```

## Task 4 — Data Visualization

Data visualization means creating charts and graphs to better understand data. Instead of just looking at numbers, graphs help us see patterns and trends quickly. Matplotlib is the most popular Python library for creating visualizations.

In this task I will create 4 types of charts: bar chart, line chart, histogram, and scatter plot.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

os.makedirs("outputs", exist_ok=True)

df = pd.read_csv("student_performance.csv")
print("Dataset loaded. Creating charts...")
```

```python
# Bar Chart - Students vs Math Marks
plt.figure(figsize=(12, 5))
plt.bar(df["Name"], df["Math"], color="steelblue")
plt.title("Students vs Math Marks")
plt.xlabel("Student Name")
plt.ylabel("Math Marks")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("outputs/bar_chart.png")
plt.show()
print("Saved: outputs/bar_chart.png")
```

The bar chart shows each student's Math marks. We can clearly see that Karthik Nair and Suresh Guptha have the highest marks while Lakshmi Devi and Divya Mehta are at the lower end.

```python
# Line Chart - Students vs Science Marks
plt.figure(figsize=(12, 5))
plt.plot(df["Name"], df["Science"], marker="o", color="green", linewidth=2)
plt.title("Students vs Science Marks")
plt.xlabel("Student Name")
plt.ylabel("Science Marks")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("outputs/line_chart.png")
plt.show()
print("Saved: outputs/line_chart.png")
```

The line chart shows how Science marks vary across students. The ups and downs in the line show that there is quite a big difference between the highest and lowest scoring students in Science.

```python
# Histogram - Distribution of Math Marks
plt.figure(figsize=(8, 5))
plt.hist(df["Math"], bins=8, color="orange", edgecolor="black")
plt.title("Distribution of Math Marks")
plt.xlabel("Math Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.savefig("outputs/histogram.png")
plt.show()
print("Saved: outputs/histogram.png")
```

The histogram shows how Math marks are spread across the class. Most students scored in the 70-90 range, with only a few students at the very low or very high end.

```python
# Scatter Plot - Math vs Science Marks
plt.figure(figsize=(7, 5))
plt.scatter(df["Math"], df["Science"], color="purple", s=80)
plt.title("Math vs Science Marks")
plt.xlabel("Math Marks")
plt.ylabel("Science Marks")
plt.tight_layout()
plt.savefig("outputs/scatter_plot.png")
plt.show()
print("Saved: outputs/scatter_plot.png")
```

The scatter plot shows the relationship between Math and Science marks. Students who scored well in Math also tend to score well in Science, which shows a positive relationship between the two subjects.

## Task 5 — Mini Project: Student Performance Data Analysis

### Introduction

This mini project is a complete analysis of student performance data. The dataset contains marks of 20 students across three subjects — Math, Science, and English — along with their attendance. The goal is to understand the overall performance of the class, find who performed best and worst, check which subject had the highest average, and determine how many students passed or failed.

### Dataset Description

The dataset has 20 student records with the following columns:
- **Student_ID** - unique ID for each student
- **Name** - student's name
- **Gender** - Male or Female
- **Age** - age of the student
- **Math** - marks in Math (out of 100)
- **Science** - marks in Science (out of 100)
- **English** - marks in English (out of 100)
- **Attendance** - attendance percentage

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

os.makedirs("outputs", exist_ok=True)

df = pd.read_csv("student_performance.csv")

print("Dataset:")
df
```

```python
# dataset info
print("Shape:", df.shape)
print()
print("Columns:", df.columns.tolist())
print()
print("Data types:")
print(df.dtypes)
print()
print("Basic Statistics:")
df.describe()
```

### Data Cleaning

```python
# checking missing values
print("Missing values:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())

# this dataset is already clean but we still check
if df.isnull().sum().sum() == 0:
    print("No missing values found.")
if df.duplicated().sum() == 0:
    print("No duplicate rows found.")
```

### Data Analysis

```python
# adding Average column
df["Average"] = ((df["Math"] + df["Science"] + df["English"]) / 3).round(2)

# adding Result column
df["Result"] = df["Average"].apply(lambda x: "Pass" if x >= 40 else "Fail")

print("Dataset with Average and Result:")
df[["Student_ID", "Name", "Math", "Science", "English", "Average", "Result"]]
```

```python
# analysis results
overall_avg = df["Average"].mean()
highest = df["Average"].max()
lowest = df["Average"].min()
pass_count = len(df[df["Result"] == "Pass"])
fail_count = len(df[df["Result"] == "Fail"])
top_student = df.loc[df["Average"].idxmax(), "Name"]
low_student = df.loc[df["Average"].idxmin(), "Name"]
avg_attendance = df["Attendance"].mean()

subject_avgs = {
    "Math": df["Math"].mean(),
    "Science": df["Science"].mean(),
    "English": df["English"].mean()
}
best_subject = max(subject_avgs, key=subject_avgs.get)

print(f"Overall class average: {overall_avg:.2f}")
print(f"Highest average: {highest}")
print(f"Lowest average: {lowest}")
print(f"Students passed: {pass_count}")
print(f"Students failed: {fail_count}")
print(f"Top student: {top_student}")
print(f"Lowest student: {low_student}")
print(f"Best subject: {best_subject} with average {subject_avgs[best_subject]:.2f}")
print(f"Average attendance: {avg_attendance:.2f}%")
```

### Visualizations

```python
# chart 1 - student average marks
plt.figure(figsize=(12, 5))
plt.bar(df["Name"], df["Average"], color="cornflowerblue")
plt.title("Student Average Marks")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("outputs/mini_project_average.png")
plt.show()
print("Saved: outputs/mini_project_average.png")
```

```python
# chart 2 - average by subject
plt.figure(figsize=(6, 5))
plt.bar(subject_avgs.keys(), subject_avgs.values(), color=["steelblue", "mediumseagreen", "tomato"])
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.tight_layout()
plt.savefig("outputs/subject_average.png")
plt.show()
print("Saved: outputs/subject_average.png")
```

```python
# chart 3 - pass vs fail pie chart
plt.figure(figsize=(5, 5))
plt.pie([pass_count, fail_count], labels=["Pass", "Fail"], autopct="%1.1f%%",
        colors=["mediumseagreen", "tomato"], startangle=90)
plt.title("Pass vs Fail Distribution")
plt.tight_layout()
plt.savefig("outputs/pass_fail.png")
plt.show()
print("Saved: outputs/pass_fail.png")
```

### Insights

Based on the analysis:

1. Most students performed well with averages above 60.
2. A few students like Lakshmi Devi and Divya Mehta need extra support.
3. There is a positive relationship between Math and Science performance.
4. Attendance is generally high across the class, which is a good sign.
5. The majority of students passed, which shows the class is doing reasonably well overall.
